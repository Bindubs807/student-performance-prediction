```python
# Cell 1: Imports and setup

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# For nicer plots
plt.style.use("seaborn-v0_8")
sns.set()

# Show plots inline (for Jupyter)
%matplotlib inline
```

```python
# Cell 2: Load the student performance dataset (local file)

from pathlib import Path

# Change YourName to your actual folder name
project_root = Path.home() / "Desktop" / "MajorProject_Bindu BS"

# Path to the dataset
data_path = project_root / "data" / "student_performance.csv"

print("Loading data from:", data_path)

df = pd.read_csv(data_path)

# Basic info
print("Shape (rows, columns):", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nFirst 5 rows:")
df.head()
```

```python
# Cell 3: Data types and missing values

print("Data types:\n")
print(df.dtypes)

print("\nMissing values per column:\n")
print(df.isna().sum())

print("\nTotal missing values:", df.isna().sum().sum())
```

```python
# Cell 4: Summary statistics for numeric columns

df.describe()
```

```python
# Save the EDA graphs for documentation

eda_path = project_root / "screenshots"

# GPA distribution
plt.figure(figsize=(8, 4))
sns.histplot(df["GPA"], bins=20, kde=True)
plt.title("Distribution of GPA")
plt.xlabel("GPA")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(eda_path / "gpa_distribution.png", dpi=300)
plt.show()

# Study time versus GPA
plt.figure(figsize=(6, 5))
sns.scatterplot(x="StudyTimeWeekly", y="GPA", data=df, alpha=0.5)
plt.title("Study Time per Week vs GPA")
plt.xlabel("StudyTimeWeekly (hours)")
plt.ylabel("GPA")
plt.tight_layout()
plt.savefig(eda_path / "study_time_vs_gpa.png", dpi=300)
plt.show()

# Absences versus GPA
plt.figure(figsize=(6, 5))
sns.scatterplot(x="Absences", y="GPA", data=df, alpha=0.5)
plt.title("Absences vs GPA")
plt.xlabel("Absences")
plt.ylabel("GPA")
plt.tight_layout()
plt.savefig(eda_path / "absences_vs_gpa.png", dpi=300)
plt.show()

print("EDA graphs saved in:", eda_path)
```
