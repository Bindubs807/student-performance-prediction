```python
# Cell 1: Import libraries and load the dataset

import pandas as pd
import numpy as np

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Project paths
project_root = Path.home() / "Desktop" / "MajorProject_Bindu BS"
data_path = project_root / "data" / "student_performance.csv"

# Load the dataset
df = pd.read_csv(data_path)

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
# Cell 2: Prepare data and train the Linear Regression model

# Remove identifier, target, and GradeClass (to avoid data leakage)
X = df.drop(columns=["StudentID", "GPA", "GradeClass"])
y = df["GPA"]

# Use the same reproducible 80/20 split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Train model
model = LinearRegression()
model.fit(X_train, y_train)

print("Model prepared and trained successfully.")
print("Features used by model:")
print(X.columns.tolist())
```

```python
# Cell 3: Create input data for a new student

new_student = pd.DataFrame({
    "Age": [17],
    "Gender": [1],
    "Ethnicity": [0],
    "ParentalEducation": [2],
    "StudyTimeWeekly": [12.0],
    "Absences": [5],
    "Tutoring": [1],
    "ParentalSupport": [3],
    "Extracurricular": [1],
    "Sports": [1],
    "Music": [0],
    "Volunteering": [0]
})

print("New Student Input Data:")
new_student
```

```python
# Cell 4: Predict GPA for the new student

predicted_gpa = model.predict(new_student)[0]

print("New Student GPA Prediction")
print("-" * 35)
print(f"Predicted GPA: {predicted_gpa:.2f} out of 4.00")

if predicted_gpa >= 3.0:
    performance_level = "Excellent"
elif predicted_gpa >= 2.0:
    performance_level = "Good"
elif predicted_gpa >= 1.0:
    performance_level = "Average"
else:
    performance_level = "Needs Improvement"

print(f"Predicted Performance Level: {performance_level}")
```
